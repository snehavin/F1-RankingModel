```python
#collecting car telemetry data for each qualifying session per driver
car_data = pd.DataFrame()
```

```python
for year in range(2019, 2025):
    for i in range(1, 25):
        try:
            session = fastf1.get_session(year, i, 'Qualifying')
            session.load()
            for driver in session.drivers:
                try:
                    data = session.car_data[driver].add_distance()
                    driver_result = session.results[session.results['DriverNumber'] == driver]
                    team = driver_result['TeamName'][0]
                    position = driver_result['Position'][0]
            
                    temp = {
                        'Driver': driver,
                        'Team': team,
                        'session_id': str(year) + '_' + str(i),
                        'season': year,
                        'Avg_Speed': data['Speed'].mean(),
                        'Max_Speed': data['Speed'].max(),
                        'Avg_Throttle': data['Throttle'].mean(),
                        'Max_Throttle': data['Throttle'].max(),
                        'Avg_RPM': data['RPM'].mean(),
                        'DRS_Activations': data['DRS'].sum(),
                        'GearChanges': data['nGear'].nunique(),
                        'Quali_Position': position
                    }
            
                    car_data = pd.concat([car_data, pd.DataFrame([temp])], ignore_index=True)
                except Exception as e:
                    print(f"Skipping driver {driver} in {session_id}: {e}")
        except Exception as e:
            print(f"Failed to load data for round {i}: {e}")
```

```python
#collecting lap time data for each qualifying session per driver
lap_data = pd.DataFrame()
```

```python
for year in range(2019, 2025):
    for i in range(1, 25):
        try:
            session = fastf1.get_session(year, i, 'Qualifying')
            session.load()
            for driver in session.drivers:
                try:
                    data = session.laps.pick_driver(driver)
            
                    avg_lap_time = data['LapTime'].mean()
                    avg_sector_1_time = data['Sector1Time'].mean()
                    avg_sector_2_time = data['Sector2Time'].mean()
                    avg_sector_3_time = data['Sector3Time'].mean()
                    
                    temp = {
                        'Driver': driver,
                        'session_id': str(year) + '_' + str(i),
                        'season': year,
                        'Avg_Lap_Time': avg_lap_time.total_seconds()if pd.notnull(avg_sector_1_time) else None,
                        'Avg_Sector_1_Time': avg_sector_1_time.total_seconds() if pd.notnull(avg_sector_1_time) else None,
                        'Avg_Sector_2_Time': avg_sector_2_time.total_seconds() if pd.notnull(avg_sector_2_time) else None,
                        'Avg_Sector_3_Time': avg_sector_3_time.total_seconds() if pd.notnull(avg_sector_3_time) else None
                    }
            
                    lap_data = pd.concat([lap_data, pd.DataFrame([temp])], ignore_index=True)
                except Exception as e:
                    print(f"Skipping driver {driver} in {session_id}: {e}")
        except Exception as e:
            print(f"Failed to load data for round {i}: {e}")
```

```python
#collecting race results - final race position for each driver
race_results = pd.DataFrame()
```

```python
#race results are not in cached data -- collecting by year to avoid rate limit errors
for i in range(1, 22):
    try:
        session = fastf1.get_session(2019, i, 'Race')
        session.load()
        for driver in session.drivers:
            try:
                driver_result = session.results[session.results['DriverNumber'] == driver]
                    
                if not driver_result.empty:
                    final_position = driver_result['Position'].iloc[0] if pd.notnull(driver_result['Position'].iloc[0]) else None
                else:
                    final_position = None
                        
                temp = {
                    'Driver': driver,
                    'session_id': '2019_' + str(i),
                    'season': 2019,
                    'Final_Race_Position': final_position
                }
        
                race_results = pd.concat([race_results, pd.DataFrame([temp])], ignore_index=True)
            except Exception as e:
                print(f"Skipping driver {driver} in {session_id}: {e}")
    except Exception as e:
        print(f"Failed to load data for round {i}: {e}")
```

```python
data = pd.merge(qualifying_data, race_results, on=['Driver', 'session_id', 'season'], how='left')
data.head()
```

```python
#XGB Ranker model does not allow String data types
#one-hot encoding Team column
encoded_data = pd.get_dummies(data, columns=['Team'])
encoded_data.head()
```

```python
encoded_data.isnull().sum()
```

```python
#dropping null values
encoded_data = encoded_data.dropna(subset=['Avg_Lap_Time', 'Avg_Sector_1_Time', 'Avg_Sector_2_Time', 'Avg_Sector_3_Time', 'Quali_Position', 'Final_Race_Position'])
```

```python
encoded_data.isnull().sum()
```

```python
encoded_data.shape
```

```python
#renaming/combining columns based on team name changes and rebranding

#Alpha Tauri, Toro Rosso, and RB became Racing Bulls
encoded_data['Team_Racing Bulls'] = encoded_data[['Team_Toro Rosso', 'Team_AlphaTauri', 'Team_RB']].any(axis=1).astype(bool)
encoded_data.drop(columns=['Team_Toro Rosso', 'Team_AlphaTauri', 'Team_RB'], inplace=True)

#Alfa Romeo became Kick Sauber
encoded_data['Team_Sauber'] = encoded_data[['Team_Alfa Romeo', 'Team_Alfa Romeo Racing', 'Team_Kick Sauber']].any(axis=1).astype(bool)
encoded_data.drop(columns=['Team_Alfa Romeo', 'Team_Alfa Romeo Racing', 'Team_Kick Sauber'], inplace=True)
encoded_data.rename(columns={'Team_Sauber':'Team_Kick Sauber'}, inplace=True)

#Renault became Alpine
encoded_data['Team_Alp'] = encoded_data[['Team_Alpine', 'Team_Renault']].any(axis=1).astype(bool)
encoded_data.drop(columns=['Team_Alpine', 'Team_Renault'], inplace=True)
encoded_data.rename(columns={'Team_Alp':'Team_Alpine'}, inplace=True)

#Racing Point became Aston Martin
encoded_data['Team_Aston'] = encoded_data[['Team_Racing Point', 'Team_Aston Martin']].any(axis=1).astype(bool)
encoded_data.drop(columns=['Team_Racing Point', 'Team_Aston Martin'], inplace=True)
encoded_data.rename(columns={'Team_Aston':'Team_Aston Martin'}, inplace=True)
```
